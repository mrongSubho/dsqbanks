#Question 1 — Titanic Fare & Survival Analysis

---



##Task 1 — Data Loading and Preparation

```python
import pandas as pd

data = pd.read_csv('/content/titanic.csv')
df = pd.DataFrame({
    'survived': data['Survived'],
    'sex': data['Sex'],
    'age': data['Age'],
    'fare': data['Fare'],
    'pclass': data['Pclass'],
    'embarked': data['Embarked']
}).dropna()
print('Shape:', df.shape)
print(df.describe())
```

##Task 2 — Grouped Central Tendency and Dispersion

```python
print(df.groupby('pclass')['fare'].agg(['mean','median']))
print()
print('Age variance:', df['age'].var())
print('Age std dev:', df['age'].std())
print('Age range:', df['age'].max() - df['age'].min())
```

##Task 3 — Survival Probability by Embarkation Port

```python
print(df.groupby('embarked')['survived'].mean().round(3))
print('Cherbourg:  {:.3f}'.format(df[df.embarked=='C']['survived'].mean()))
print('Queenstown: {:.3f}'.format(df[df.embarked=='Q']['survived'].mean()))
print('Southampton:{:.3f}'.format(df[df.embarked=='S']['survived'].mean()))
```

##Task 4 — One-Way ANOVA: Fare Across Passenger Classes

```python
from scipy.stats import f_oneway
c1 = df[df.pclass==1]['fare']
c2 = df[df.pclass==2]['fare']
c3 = df[df.pclass==3]['fare']
f, p = f_oneway(c1, c2, c3)
print('F-statistic:', round(f, 4))
print('p-value:', '{:.2e}'.format(p))
print('Significant:', 'Yes' if p < 0.05 else 'No')
```
