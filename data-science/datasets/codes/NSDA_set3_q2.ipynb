##Task 1 — Load and Explore Data

```python
import seaborn as sns
tips = sns.load_dataset('tips')
print('Shape:', tips.shape)
print(tips.head())
print(tips.dtypes)
print(tips.isnull().sum())
print(tips.describe())
```

##Task 2 — Tip Percentage and Central Tendency

```python
import matplotlib.pyplot as plt
tips['tip_pct'] = tips['tip'] / tips['total_bill'] * 100
mean_tip = tips['tip_pct'].mean()
median_tip = tips['tip_pct'].median()
mode_tip = tips['tip_pct'].mode()[0]
print(f'Mean: {mean_tip:.2f}%')
print(f'Median: {median_tip:.2f}%')
print(f'Mode: {mode_tip:.2f}%')
tips['tip_pct'].plot(kind='hist', bins=10, density=True, alpha=0.6, figsize=(8,5))
tips['tip_pct'].plot(kind='kde', color='red')
plt.xlabel('Tip Percentage'); plt.title('Distribution of Tip Percentage')
plt.show()
```

##Task 3 — Hypothesis Testing: Smokers vs Non-Smokers

```python
from scipy.stats import ttest_ind
smoker = tips[tips.smoker=='Yes']['tip_pct']
non_smoker = tips[tips.smoker=='No']['tip_pct']
t, p = ttest_ind(smoker, non_smoker)
print(f'Smoker mean: {smoker.mean():.2f}% (n={len(smoker)})')
print(f'Non-smoker mean: {non_smoker.mean():.2f}% (n={len(non_smoker)})')
print(f't-statistic: {t:.3f}, p-value: {p:.4f}')
print('Significant difference?', 'Yes' if p < 0.05 else 'No')
```

##Task 4 — ANOVA: Tip Percentage Across Days

```python
from scipy.stats import f_oneway
thur = tips[tips.day=='Thur']['tip_pct']
fri = tips[tips.day=='Fri']['tip_pct']
sat = tips[tips.day=='Sat']['tip_pct']
sun = tips[tips.day=='Sun']['tip_pct']
f, p = f_oneway(thur, fri, sat, sun)
print(f'ANOVA: F={f:.3f}, p={p:.4f}')
print('Significant?', 'Yes' if p < 0.05 else 'No')
print()
print('Group means:')
print(tips.groupby('day')['tip_pct'].mean().round(2))
```
