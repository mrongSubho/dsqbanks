#Question 2 — Iris Dataset Analysis

---



##Task 1 — Data Loading and Exploration

```python
import seaborn as sns
iris = sns.load_dataset('iris')
print(iris.shape)
print(iris.dtypes)
print(iris.isnull().sum())
print(iris['species'].value_counts())
```

##Task 2 — Per-Species Summary Statistics

```python
import matplotlib.pyplot as plt
means = iris.groupby('species').mean()
means.plot(kind='bar', figsize=(8,5))
plt.title('Mean Feature Values by Species')
plt.ylabel('Centimeters')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

##Task 3 — Distribution Analysis

```python
fig, axes = plt.subplots(2, 2, figsize=(10,8))
features = ['sepal_length','sepal_width','petal_length','petal_width']
for i, f in enumerate(features):
    ax = axes[i//2, i%2]
    for sp in iris['species'].unique():
        iris[iris.species==sp][f].plot(kind='kde', ax=ax, label=sp)
    ax.set_title(f)
plt.tight_layout()
plt.show()
iris.boxplot(by='species', figsize=(10,6))
plt.tight_layout()
plt.show()
print('Petal length and petal width best separate the three species, with setosa having distinctly smaller values.')
```

##Task 4 — Correlation Heatmap

```python
import seaborn as sns
plt.figure(figsize=(6,5))
sns.heatmap(iris[['sepal_length','sepal_width','petal_length','petal_width']].corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Iris Feature Correlations')
plt.tight_layout()
plt.show()
print('Strongest correlation: petal_length vs petal_width =', iris['petal_length'].corr(iris['petal_width']).round(3))
```
