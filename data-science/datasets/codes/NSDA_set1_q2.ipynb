#Question 2 — Data Visualization

##Task 1 — Data Loading and Preparation


```python
import pandas as pd
data = pd.read_csv('/content/titanic.csv')
df = data[['Survived', 'Pclass', 'Sex', 'Age', 'Fare']].dropna()
print('Shape:', df.shape)
df.head()
```

##Task 2 — Survival Report


```python
import matplotlib.pyplot as plt
surv_by_class = df.groupby('Pclass')['Survived'].sum()
surv_by_class.plot(kind='bar', title='Survival by Class')
plt.show()

surv_by_sex = df.groupby('Sex')['Survived'].sum()
surv_by_sex.plot(kind='bar', title='Survival by Sex')
plt.show()
```

##Task 3 — Distribution of Age


```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(df['Age'], kde=True)
plt.title('Age distribution with KDE')
plt.show()

sns.boxplot(x=df['Age'])
plt.title('Age boxplot')
plt.show()
```

##Task 4 — Correlation Heatmap


```python
import seaborn as sns
import matplotlib.pyplot as plt
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

print('Strongest correlations:')
print(corr.unstack().sort_values())
```
