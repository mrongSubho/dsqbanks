#Question 3 — Machine Learning

##Task 1 — Data Loading and Pre-processing


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('/content/student_performance.csv')
df = df.dropna()
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
df['ParentEducation'] = le.fit_transform(df['ParentEducation'])
df['InternetAccess'] = le.fit_transform(df['InternetAccess'])
df['Result'] = (df['FinalScore'] >= 50).astype(int)
scaler = StandardScaler()
num_cols = ['StudyHoursPerWeek', 'AttendanceRate', 'FinalScore']
df[num_cols] = scaler.fit_transform(df[num_cols])
df.head()
```

##Task 2 — Model Training


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

features = ['Gender', 'ParentEducation', 'StudyHoursPerWeek', 'AttendanceRate', 'InternetAccess']
X = df[features]
y = df['Result']
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

##Task 3 — Model Evaluation


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

y_pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:', recall_score(y_test, y_pred))
print('Confusion Matrix:\n', confusion_matrix(y_test, y_pred))
```

##Task 4 — Save the Model


```python
import pickle
with open('model.pkl', 'wb') as f:
    pickle.dump(model, f)

with open('model.pkl', 'rb') as f:
    loaded = pickle.load(f)
print('Sample prediction:', loaded.predict(X_test.iloc[[0]]))
```
